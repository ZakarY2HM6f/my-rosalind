# [Introduction to the Bioinformatics Armory](https://rosalind.info/problems/ini/)

**Given**: A DNA string s of length at most 1000 bp.

**Return**: Four integers (separated by spaces) representing the respective number of times that the symbols 'A', 'C', 'G', and 'T' occur in s. Note: You must provide your answer in the format shown in the sample output below.

```python
dataset = "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"
```

```python
dataset = Path("./datasets/rosalind_ini.txt").read_text()
```

```python
from Bio.Seq import Seq

seq = Seq(dataset)
result = [
    seq.count("A"),
    seq.count("C"),
    seq.count("G"),
    seq.count("T"),
]
print(" ".join([str(i) for i in result]))
```

# [GenBank Introduction](https://rosalind.info/problems/gbk/)

**Given**: A genus name, followed by two dates in YYYY/M/D format.

**Return**: The number of Nucleotide GenBank entries for the given genus that were published between the dates specified.

```python
dataset = """Anthoxanthum
2003/7/25
2005/12/27"""
```

```python
dataset = Path("./datasets/rosalind_gbk.txt").read_text()
```

```python
lines = dataset.splitlines()

name = lines[0]
mindate = lines[1]
maxdate = lines[2]
```

```python
from Bio import Entrez

Entrez.email = "s2775530@ed.ac.uk"

handle = Entrez.esearch(
    db="nucleotide",
    term=f"{name}[Organism]",
    datetype="pdat",
    mindate=mindate,
    maxdate=maxdate,
)
record = Entrez.read(handle)
record["Count"]
```

# [Data Formats](https://rosalind.info/problems/frmt/)

**Given**: A collection of n (n≤10) GenBank entry IDs.

**Return**: The shortest of the strings associated with the IDs in FASTA format.

```python
dataset = """FJ817486 JX069768 JX469983"""
```

```python
dataset = Path("./datasets/rosalind_frmt.txt").read_text()
```

```python
ids = dataset.split(" ")
```

```python
from Bio import Entrez
from Bio import SeqIO

Entrez.email = "s2775530@ed.ac.uk"

handle = Entrez.efetch(
    db="nucleotide",
    id=ids,
    rettype="fasta",
)
records = SeqIO.parse(handle, "fasta")
records = sorted(records, key=lambda x: len(x.seq))

print(records[0].format("fasta"))
```
